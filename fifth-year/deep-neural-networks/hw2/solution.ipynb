## Overview
This homework has two parts.

The first part of this homework is to implement [Grad-CAM](https://arxiv.org/pdf/1610.02391) – a method for producing saliency maps (heatmaps of regions that are most relevant to a model) that uses both activations and gradients of the feature maps of a convolutional layer. Feature maps of deeper convolutional layers represent more high-level features, while preserving rough spatial structure, which makes them a good candidate for explaining a model's output.

The second part of this homework is to use SAM for image segmentation without user input.
SAM [(Segment Anything Model v1)](https://arxiv.org/pdf/2304.02643) is a popular family of open-weight models for image segmentation (based on the vision transformer ViT and CLIP).
The model takes as input an image to be segmented and additionaly bounding boxes, point coordinates etc. clarifying the object of interest to be segmented. It can output many proposed segmentations of many objects on one image. Your task will be to find appropriate point coordinates automatically, so that SAM can be used with just an image input to segment particular objects.

```python
class SyntheticData[T](Dataset):
    """A small synthetic segmentation dataset.

    It is a sequence dataset of 5000 tuples (image, class, mask), where:
    - image: before transformation, an RGB PIL Image.
    - class: int 0..4, the label index.
    - mask:  numpy array of dtype=bool, shape (H, W), same size as image.
    """

    CLASSES: Final[tuple[str, ...]] = ("circle", "square", "triangle", "star", "diamond")

    def __init__(
        self,
        path: Path,
        split: Literal["train", "val", "test"],
        transform: Callable[[PIL.Image.Image], T],
    ) -> None:
        self.path = path
        self.transform = transform
        self.split = split

        self.class_to_idx = {class_name: i for i, class_name in enumerate(self.CLASSES)}

        images = sorted((self.path / "images" / self.split).glob("*.png"))
        masks = sorted((self.path / "masks" / self.split).glob("*.png"))
        labels = sorted((self.path / "labels" / self.split).glob("*.txt"))

        assert images, f"No images found in {self.path / 'images' / self.split}"
        assert len(images) == len(masks) == len(labels), (
            "Number of images, masks, and labels must be the same"
        )
        assert [p.stem for p in images] == [p.stem for p in masks] == [p.stem for p in labels], (
            "Image/mask/label filename mismatch."
        )
        self.image_names = [p.stem for p in images]

    def __len__(self) -> int:
        return len(self.image_names)

    def __getitem__(self, idx) -> tuple[T, int, np.ndarray]:
        name = self.image_names[idx]
        img_path = self.path / "images" / self.split / f"{name}.png"
        mask_path = self.path / "masks" / self.split / f"{name}.png"
        label_path = self.path / "labels" / self.split / f"{name}.txt"

        img = PIL.Image.open(img_path).convert("RGB")
        mask = np.array(PIL.Image.open(mask_path), dtype=bool)
        label = self.class_to_idx[label_path.read_text().strip()]

        img_transformed: T = self.transform(img)

        return img_transformed, label, mask


def show_image_row(
    image_dict: dict[str, PIL.Image.Image | np.ndarray | Tensor], size: float = 3.0
) -> None:
    n = len(image_dict)
    _, axs = plt.subplots(1, n, figsize=(size * n, size), constrained_layout=True, squeeze=True)
    if n == 1:
        axs = [axs]
    for ax, (title, img) in zip(axs, image_dict.items(), strict=True):
        ax.imshow(img)
        ax.set_title(title)
        ax.axis("off")
    plt.show()
```

## Training a classifier [do not modify]

```python
def train(
    model: nn.Module,
    train_loader: torch.utils.data.DataLoader,
    val_loader: torch.utils.data.DataLoader,
    test_loader: torch.utils.data.DataLoader,
    epochs: int = 5,
    **optimizer_kwargs: Any,
) -> None:
    print(f"🚀 Training CNN for {epochs} epochs...")
    optimizer = optim.AdamW(model.parameters(), **optimizer_kwargs)

    for epoch in range(epochs):
        _train_epoch(model, train_loader, optimizer, desc=f"Epoch {epoch + 1}/{epochs} training  ")
        val_metrics = evaluate(model, val_loader, desc=f"Epoch {epoch + 1}/{epochs} validation")
        print(
            f"Epoch {epoch + 1}/{epochs} — "
            + f"val loss: {val_metrics['loss']:.3f}, val acc: {val_metrics['accuracy']:.1%}"
        )

    test_metrics = evaluate(model, test_loader, desc="Test Evaluation")
    print(
        "✅ Model training complete: "
        + f"Test loss: {test_metrics['loss']:.3f}, test acc: {test_metrics['accuracy']:.1%}"
    )


def _train_epoch(
    model: nn.Module,
    dataloader: torch.utils.data.DataLoader,
    optimizer: optim.Optimizer,
    desc: str,
) -> dict[str, float]:
    model.train()
    device = next(model.parameters()).device
    total_loss = 0
    total_correct = 0
    total_samples = 0
    progress_bar = tqdm(dataloader, desc=desc)
    for imgs, labels, _ in progress_bar:
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = nn.CrossEntropyLoss()(outputs, labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

        _, predicted = torch.max(outputs.data, dim=1)
        total_samples += labels.shape[0]
        total_correct += (predicted == labels).sum().item()

        progress_bar.set_postfix(
            train_loss=f"{total_loss / (total_samples / labels.shape[0]):.3f}",
            train_acc=f"{total_correct / total_samples:.1%}",
        )

    return {"loss": total_loss / len(dataloader), "accuracy": total_correct / total_samples}


def evaluate(
    model: nn.Module, dataloader: torch.utils.data.DataLoader, desc: str
) -> dict[str, float]:
    model.eval()
    device = next(model.parameters()).device
    total_loss = 0
    total_correct = 0
    total_samples = 0
    with torch.no_grad():
        for imgs, labels, _mask in tqdm(dataloader, desc=desc):
            imgs, labels = imgs.to(device), labels.to(device)
            outputs = model(imgs)
            loss = nn.CrossEntropyLoss()(outputs, labels)
            total_loss += loss.item()
            _, predicted = torch.max(outputs.data, dim=1)
            total_samples += labels.shape[0]
            total_correct += (predicted == labels).sum().item()

    return {"loss": total_loss / len(dataloader), "accuracy": total_correct / total_samples}


class DataloaderArgs(TypedDict, total=False):
    batch_size: int
    shuffle: bool
    num_workers: int
    pin_memory: bool
```

## GradCAM implementation (add your code)

```python
class GradCAM:
    """
    Pure PyTorch implementation of Grad-CAM.

    Usage:
        grad_cam = GradCAM(model=model, target_layers=[layer1, layer2])
        grayscale_cam = grad_cam(input_tensor, targets=[class_id])
    """

    def __init__(self, model: nn.Module, target_layers: Iterable[nn.Module]) -> None:
        self.model = model
        self.target_layers = list(target_layers)
        self._handles = []
        self._activations = [None] * len(self.target_layers) 
        self._gradients = [None] * len(self.target_layers)

        # record activations and gradients
        for i, layer in enumerate(self.target_layers):
            self._handles.append(layer.register_forward_hook(self._hook(i)))

    def _hook(self, i: int):
        def set_gradient(i, g):
            self._gradients[i] = g
        def set_activation_and_gradient_hook(module, args, out):
            self._activations[i] = out
            out.register_hook(lambda g: set_gradient(i, g))
        return set_activation_and_gradient_hook

    def close(self) -> None:
        for handle in self._handles:
            handle.remove()
        self._handles.clear()

    def __call__(self, input_tensor: Tensor, targets: Iterable[int] | None = None) -> np.ndarray:
        """
        Returns: numpy array of shape (B, H, W), values 0..1.
        """
        B, C, H, W = input_tensor.shape
        self.model.eval()

        # clear activation and gradients 
        self._activations = [None] * len(self.target_layers)
        self._gradients = [None] * len(self.target_layers)

        # compute classes 
        outputs = self.model(input_tensor)

        # turn targets into a tensor; use the most likely answer if not provided 
        if targets is None:
            target_tensor = outputs.argmax(dim=1)
        else:
            target_tensor = torch.tensor(targets, device=outputs.device, dtype=torch.long)
            assert target_tensor.shape[0] == B

        self.model.zero_grad(set_to_none=True)
        score = outputs[torch.arange(B), target_tensor].sum()
        # fill the gradients
        score.backward()

        cams_per_layer: list[Tensor] = []
        for activation, gradient in zip(self._activations, self._gradients, strict=True):
            weights = gradient.mean(dim=(2, 3), keepdim=True)  # alpha_b_k
            cam = (weights * activation).sum(dim=1, keepdim=True)
            cam = F.relu(cam)

            # we need to have the same resolution for all layers
            cam = F.interpolate(cam, size=(H, W), mode="bilinear", align_corners=False)
            cam = cam.squeeze(1)

            # normalize independently per layer
            cam_min = cam.amin(dim=(1, 2), keepdim=True)
            cam_max = cam.amax(dim=(1, 2), keepdim=True)
            cam = (cam - cam_min) / (cam_max - cam_min + 1e-8)
            cams_per_layer.append(cam)

        cams = torch.stack(cams_per_layer, dim=0).mean(dim=0)
        return cams.detach().cpu().numpy()
```

```python
def heatmap_to_rgb_image(
    heatmap: np.ndarray, min: float | None = None, max: float | None = None
) -> PIL.Image.Image:
    """
    Converts a single-channel heatmap to an RGB pillow image using a colormap.

    Args:
    - heatmap: shape (H, W), will be normalized by mapping min..max to 0..1.
    - min: minimum value for normalization, defaults to heatmap.min().
    - max: maximum value for normalization, defaults to heatmap.max()
    """
    heatmap = heatmap.astype(np.float32)
    if min is None:
        min = heatmap.min()
    if max is None:
        max = heatmap.max()
    heatmap = (heatmap - min) / (max - min + 1e-8)
    heatmap_uint8 = (np.clip(heatmap, 0.0, 1.0) * 255).astype(np.uint8)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)
    return PIL.Image.fromarray(heatmap_color)
```

```python
class BasicSamPipeline:
    def __call__(self, images: Tensor) -> Tensor:
        """
        Input: normalized images, shape (B, C=3, H, W).
        Output: masks tensor of shape (B, H, W), dtype=bool.
        """
        B, C, H, W = images.shape
        # The basic pipeline always uses a single center point for each image.
        point_coords = np.array([[(W // 2, H // 2)] for _ in range(B)])

        # The basic pipeline always uses a single foreground point, no background points.
        point_labels = np.array([[1] for _ in range(B)], dtype=np.int64)

        return self.segment(images, point_coords, point_labels)

    def segment(
        self, images: Tensor, point_coords: np.ndarray, point_labels: np.ndarray
    ) -> Tensor:
        """
        Args:
        - images: normalized images, shape (B, C=3, H, W).
        - point_coords: point coordinates within each image, shape (B, num_points, 2), format (x,y).
            Note the format is not (h,w)=(y,x), but (x,y)!
        - point_labels: point labels, shape (B, num_points), dtype int64.
            Label 1 is foreground (should be in mask), 0 is background (shouldn't be in mask).

        Returns: segmentation masks, shape (B, H, W), dtype=bool.
        """
        B, C, H, W = images.shape
        assert C == 3, f"Expected images.shape=(B, C=3, H, W), got: {images.shape}"
        num_points = point_coords.shape[1]
        assert point_coords.shape == (B, num_points, 2), f"Expected point_coords.shape=({B=}, num_points, 2), got: {point_coords.shape}"
        assert point_labels.shape == (B, num_points), f"Expected point_labels.shape=({B=}, num_points), got: {point_labels.shape}"

        results = list[Tensor]()
        for image, pt_coords, pt_labels in zip(images, point_coords, point_labels, strict=True):
            sam_predictor.set_image(np.array(inverse_transform(image)))
            masks, scores, _logits = sam_predictor.predict(
                point_coords=pt_coords, point_labels=pt_labels, multimask_output=True
            )
            best_mask = masks[np.argmax(scores)]
            results.append(torch.tensor(best_mask, dtype=torch.bool))
        return torch.stack(results)
```

```python
class GCSamPipeline(BasicSamPipeline):
    """Base class for SAM pipelines that use gradcam.
    """

    def __init__(self, grad_cam: GradCAM) -> None:
        super().__init__()
        self.grad_cam = grad_cam

    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        """
        Retrieve points from a cam heatmap
        Returns:
            point_coords: (P, 2) in (x, y) format
            point_labels:  (P,) with 0 for bg 1 for fg
        """
        raise NotImplementedError

    def __call__(self, images: Tensor) -> Tensor:
        with torch.enable_grad():
            cams = self.grad_cam(images.to(device), targets=None)  # (B, H, W)

        # Convert each heatmap to point prompts
        coords_list: list[np.ndarray] = []
        labels_list: list[np.ndarray] = []
        for cam_i in cams:
            coords_i, labels_i = self.get_points(cam_i)
            coords_list.append(coords_i)
            labels_list.append(labels_i)

        point_coords = np.stack(coords_list, axis=0)
        point_labels = np.stack(labels_list, axis=0)
        return self.segment(images, point_coords, point_labels)


class ForegroundGCSamPipeline(GCSamPipeline):
    """select most active points as fg"""
    def __init__(self, grad_cam: GradCAM, n_fg_points: int = 1): 
        super().__init__(grad_cam)
        self.n_fg_points = n_fg_points
    
    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        H, W = cam.shape
        P = self.n_fg_points
        flat = cam.reshape(-1)
        idxs = np.argpartition(-flat, kth=P - 1)[:P]
        idxs = idxs[np.argsort(-flat[idxs])]

        ys = idxs // W
        xs = idxs % W
        point_coords = np.stack([xs, ys], axis=1)
        point_labels = np.ones(P, dtype=np.int64)
        return point_coords, point_labels


class BackgroundGCSamPipeline(GCSamPipeline):
    """select least active points as bg"""
    
    def __init__(self, grad_cam: GradCAM, n_bg_points: int = 1):
        super().__init__(grad_cam)
        self.n_bg_points = n_bg_points

    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        H, W = cam.shape
        P = self.n_bg_points

        flat = cam.ravel()
        idxs = np.argpartition(flat, P - 1)[:P]
        idxs = idxs[np.argsort(flat[idxs])]

        xs = idxs % W
        ys = idxs // W
        point_coords = np.stack([xs, ys], axis=1)
        point_labels = np.zeros(P, dtype=np.int64)
        return point_coords, point_labels

    
class CompositionGCSamPipeline(GCSamPipeline):
    """compose a few pipelines together (ie stack the points)"""
    
    def __init__(self, grad_cam: GradCAM, pipelines: list[GCSamPipeline]): 
        super().__init__(grad_cam)
        self.pipelines = pipelines
    
    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        coords_parts = []
        labels_parts = []

        for pipe in self.pipelines:
            coords, labels = pipe.get_points(cam)
            coords_parts.append(coords)
            labels_parts.append(labels)

        point_coords = np.concatenate(coords_parts, axis=0)
        point_labels = np.concatenate(labels_parts, axis=0)
        return point_coords, point_labels
```

```python
from dataclasses import dataclass
import PIL.ImageDraw


@dataclass
class PipelineMetrics:
    name: str
    mean_iou: float
    fg_hit_rate: float
    fg_distance: float
    bg_hit_rate: float | None = None
    bg_distance: float | None = None


class SamPipelineEvaluator:
    
    def __init__(
        self,
        *,
        model: nn.Module,
        target_layers: list[nn.Module],
        eval_loader: DataLoader,
        example_loader: DataLoader,
        max_batches: int | None = None,
    ) -> None:
        self.grad_cam = GradCAM(model=model, target_layers=target_layers)
        self.eval_loader = eval_loader
        self.example_loader = example_loader
        self.max_batches = max_batches

    def evaluate(self, name: str, pipeline: GCSamPipeline) -> PipelineMetrics:
        fg_hit_rate, fg_distance, bg_hit_rate, bg_distance = self._evaluate_points(pipeline)
        mean_iou = self._evaluate_iou(pipeline)
        
        return PipelineMetrics(
            name=name,
            mean_iou=mean_iou,
            fg_hit_rate=fg_hit_rate,
            fg_distance=fg_distance,
            bg_hit_rate=bg_hit_rate,
            bg_distance=bg_distance,
        )

    def _evaluate_points(self, pipeline: GCSamPipeline) -> tuple[float, float, float | None, float | None]:
        """Evaluate point prompt quality: hit rate and distance from mask center."""
        fg_hits = 0
        fg_total = 0
        bg_hits = 0
        bg_total = 0
        fg_dist_sum = 0.0 
        bg_dist_sum = 0.0

        for b_idx, (images, _, gt_masks) in enumerate(self.eval_loader):
            if self.max_batches is not None and b_idx >= self.max_batches:
                break
            
            gt_masks = gt_masks.cpu().numpy()
            
            with torch.enable_grad():
                cams = pipeline.grad_cam(images.to(device), targets=None)

            for i in range(gt_masks.shape[0]):
                mask = gt_masks[i]
                cy, cx = scipy.ndimage.center_of_mass(mask.astype(np.float32))
                coords, labels = pipeline.get_points(cams[i])
                
                for (x, y), lbl in zip(coords, labels, strict=True):
                    x_i = int(np.clip(x, 0, mask.shape[1] - 1))
                    y_i = int(np.clip(y, 0, mask.shape[0] - 1))

                    inside = bool(mask[y_i, x_i])
                    dist = float(((x_i - cx) ** 2 + (y_i - cy) ** 2) ** 0.5)

                    if lbl == 1:  # foreground point
                        fg_total += 1
                        fg_hits += int(inside)
                        fg_dist_sum += dist
                    else:  # background point
                        bg_total += 1
                        bg_hits += int(not inside) 
                        bg_dist_sum += dist

        fg_hit_rate = (fg_hits / fg_total) if fg_total else float("nan")
        fg_distance = (fg_dist_sum / fg_total) if fg_total else float("nan")
        bg_hit_rate = (bg_hits / bg_total) if bg_total else None
        bg_distance = (bg_dist_sum / bg_total) if bg_total else None
        
        return fg_hit_rate, fg_distance, bg_hit_rate, bg_distance

    def _evaluate_iou(self, pipeline: BasicSamPipeline) -> float:
        """Evaluate segmentation quality using Intersection over Union."""
        ious: list[float] = []
        
        for b_idx, (images, _labels, gt_masks) in enumerate(tqdm(self.eval_loader)):
            if self.max_batches is not None and b_idx >= self.max_batches:
                break
            
            preds = pipeline(images).cpu().numpy()
            gt_masks = gt_masks.cpu().numpy()
            
            intersection = np.logical_and(preds, gt_masks).sum(axis=(1, 2))
            union = np.logical_or(preds, gt_masks).sum(axis=(1, 2))
            batch_iou = np.where(union > 0, intersection / union, 1.0)
            ious.append(batch_iou.mean())
        
        return np.mean(ious) if ious else float("nan")

    def show_examples(
        self,
        pipeline: GCSamPipeline,
        name: str,
        num_examples: int = 3,
    ) -> None:
        """Visualize pipeline results with GradCAM overlay and predicted masks."""
        images, _labels, gt_masks = next(iter(self.example_loader))
        indices = list(range(min(num_examples, images.shape[0])))
        images = images[indices]
        gt_masks = gt_masks[indices]

        preds = pipeline(images).cpu()
        with torch.enable_grad():
            cams = pipeline.grad_cam(images.to(device), targets=None)

        for image, pred, gt_mask, cam in zip(images, preds, gt_masks, cams, strict=True):
            coords, labels = pipeline.get_points(cam)
            
            # Create overlay with point annotations
            base = inverse_transform(image).copy()
            heatmap = heatmap_to_rgb_image(np.asarray(cam), 0, 1)
            overlay = PIL.Image.blend(base, heatmap, alpha=0.35)
            draw = PIL.ImageDraw.Draw(overlay)

            # draw points
            W, H = overlay.size
            radius = 1
            for (x, y), lbl in zip(coords, labels, strict=True):
                x_i = int(np.clip(round(x), 0, W - 1))
                y_i = int(np.clip(round(y), 0, H - 1))
                color = "lime" if lbl == 1 else "red"
                draw.ellipse((x_i - radius, y_i - radius, x_i + radius, y_i + radius), 
                           fill=color, outline="black")

            show_image_row({
                "Image + GradCAM overlay": overlay,
                f"{name} predicted mask": pred,
                "Ground-truth mask": gt_mask,
            })

    def print_report(self, metrics: PipelineMetrics) -> None:
        print("\n" + "=" * 80)
        print(metrics.name)
        print(f"FG hit rate: {metrics.fg_hit_rate:.1%}")
        print(f"FG distance: {metrics.fg_distance:.2f} px")
        if metrics.bg_hit_rate is not None:
            print(f"BG hit rate (outside mask): {metrics.bg_hit_rate:.1%}")
            print(f"BG distance: {metrics.bg_distance:.2f} px")
        print(f"Mean IoU: {metrics.mean_iou:.1%}")

    def print_summary(self, results: list[PipelineMetrics]) -> None:
        headers = ["Configuration", "Mean IoU", "FG hit", "FG dist", "BG hit", "BG dist"]
        rows: list[list[str]] = []
        
        for r in results:
            rows.append([
                r.name,
                f"{r.mean_iou:.1%}",
                f"{r.fg_hit_rate:.1%}",
                f"{r.fg_distance:.2f}",
                "" if r.bg_hit_rate is None else f"{r.bg_hit_rate:.1%}",
                "" if r.bg_distance is None else f"{r.bg_distance:.2f}",
            ])

        # Calculate column widths
        col_widths = [len(h) for h in headers]
        for row in rows:
            for i, val in enumerate(row):
                col_widths[i] = max(col_widths[i], len(val))

        # Print table
        def format_row(values: list[str]) -> str:
            return " | ".join(v.ljust(col_widths[i]) for i, v in enumerate(values))

        print("\n" + format_row(headers))
        print("-+-".join("-" * w for w in col_widths))
        for row in rows:
            print(format_row(row))
```

```python
class GridBackgroundGCSamPipeline(GCSamPipeline):
    """divide into grids; select a bg point in each"""
    
    def __init__(self, grad_cam: GradCAM, *, rows: int = 4, cols: int = 4):
        super().__init__(grad_cam)
        self.rows = rows
        self.cols = cols

    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        H, W = cam.shape

        coords: list[tuple[int, int]] = []
        for r in range(self.rows):
            y0 = (r * H) // self.rows
            y1 = ((r + 1) * H) // self.rows
            for c in range(self.cols):
                x0 = (c * W) // self.cols
                x1 = ((c + 1) * W) // self.cols
                tile = cam[y0:y1, x0:x1]
                assert tile.size > 0
                idx = int(tile.argmin())
                ty, tx = np.unravel_index(idx, tile.shape)
                coords.append((int(x0 + tx), int(y0 + ty)))

        point_coords = np.array(coords)
        point_labels = np.zeros(len(coords), dtype=np.int64)
        return point_coords, point_labels
```

```python
class LocalCircleBackgroundGCSamPipeline(GCSamPipeline):
    """select bg points in a circle around the most heated point"""
    
    def __init__(self, grad_cam: GradCAM, radius: int = 20, n_bg_points: int = 3, delta: float = 10.0):
        super().__init__(grad_cam)
        self.radius = radius
        self.n_bg_points = n_bg_points
        self.delta = delta  # minimum distance between points for diversity

    def get_points(self, cam: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
        assert cam.ndim == 2
        H, W = cam.shape
        r = int(self.radius)
        P = int(self.n_bg_points)
        assert r >= 1 

        # Find the peak activation (likely object center)
        flat = cam.ravel()
        fg_idx = int(flat.argmax())
        fx, fy = fg_idx % W, fg_idx // W

        # Define circular region around peak
        ys, xs = np.ogrid[:H, :W]
        circle = (xs - fx) ** 2 + (ys - fy) ** 2 <= r * r
        ys_c, xs_c = np.where(circle)
        vals = cam[ys_c, xs_c]

        # Select diverse low-activation points within the circle
        order = np.argsort(vals)  # lowest first
        picked: list[int] = []
        d2 = float(self.delta) ** 2

        for j in order:
            if len(picked) == P:
                break
            x, y = int(xs_c[j]), int(ys_c[j])
            # Check if point is far enough from already picked points
            if all((x - int(xs_c[k])) ** 2 + (y - int(ys_c[k])) ** 2 >= d2 for k in picked):
                picked.append(j)

        assert len(picked) == P, "Not enough diverse BG points; reduce delta or increase radius"

        xs_bg = xs_c[picked]
        ys_bg = ys_c[picked]
        point_coords = np.stack([xs_bg, ys_bg], axis=1)
        point_labels = np.zeros(P, dtype=np.int64)
        return point_coords, point_labels
```
